```python
import json
import matplotlib.pyplot as plt
from pathlib import Path
import numpy as np
import seaborn as sns
```

```python
results_dir = Path("artifacts/metrics/")
ending = 'val'

files = {
    "Baseline": results_dir / f"baseline_{ending}.json",
    "Medium": results_dir / f"medium_{ending}.json",
    "Advanced": results_dir / f"advanced_{ending}.json",
}

labels = []
```

```python
def plot_metric(metric_name, values):
    plt.figure(figsize=(7,5))

    x = np.arange(len(labels))  # positions: 0,1,2

    bars = plt.bar(x, values)

    # put model names on x axis
    plt.xticks(x, labels)

    # title and axis
    plt.title(f"Model Comparison - {metric_name}")
    plt.ylabel(metric_name)

    # ---- add value labels above each bar ----
    for bar in bars:
        height = bar.get_height()
        plt.text(
            bar.get_x() + bar.get_width()/2,
            height + 0.005,
            f'{height:.3f}',
            ha='center',
            va='bottom',
            fontsize=10
        )

    plt.ylim(0, 1)  # metrics are between 0 and 1
    plt.tight_layout()
    plt.savefig(f"{metric_name}_comparison.png", dpi=300)
    plt.close()

def plot_confusion_matrix(cm, model_name):

    cm = np.array(cm)

    plt.figure(figsize=(7,6))

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        cbar=True
    )

    plt.title(f"Confusion Matrix - {model_name}")
    plt.xlabel("Predicted label")
    plt.ylabel("True label")

    plt.tight_layout()
    plt.savefig(f"confusion_{model_name}.png", dpi=300)
    plt.close()
```

```python
metrics = {
    "accuracy": [],
    "f1_weighted": [],
    "f1_macro": []
}

matrix = []

for name, path in files.items():
    with open(path, "r") as f:
        data = json.load(f)
    labels.append(name)

    metrics["accuracy"].append(data["accuracy"])
    metrics["f1_weighted"].append(data["f1_weighted"])
    metrics["f1_macro"].append(data["f1_macro"])
    matrix.append(data['confusion_matrix'])
```

```python
for metric, values in metrics.items():
    plot_metric(metric, values)

print("Pie charts generated!")
```

```python
models = ["Baseline", "Medium", "Advanced"]
for i, cm in enumerate(matrix):
    plot_confusion_matrix(cm, models[i])

print("Confusion matrices created!")
```
